```python
from ultralytics import YOLO
model = YOLO("yolo11x.pt") 

data_path = 'ships-aerial-images/data.yaml'

train_params = {
    'epochs': 20,
    'batch': 8,
    'imgsz': 640,
    'lr0': 0.01,
    'lrf': 0.01,
    'momentum': 0.937,
    'weight_decay': 0.0005,
    'optimizer': 'Adam',
    'device': '0,1',
    'pretrained': True,
}
model.train(data=data_path, **train_params)
```
